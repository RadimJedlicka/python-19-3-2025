### **Ukázka #03**

Rozdílná pravidla při vytváření a zapisování slovníků

```python
muj_slovnik_1 = dict("jmeno": "Matous", "prijmeni": "Nezname")
```

```python
muj_slovnik_1 = {
    "jmeno": "Matous",
    "prijmeni": "Nezname"
    }
```

```python
print(muj_slovnik_1)
```

```python
muj_slovnik_2 = dict(jmeno="Matous", prijmeni="Nezname")
```

```python
print(muj_slovnik_2)
```

```python
pismena = ["a", "b" ,"c", "d"]
vyskyt = [1, 10, 5, 0]
muj_slovnik_3 = {}.fromkeys(pismena, 0)
```

```python
print(muj_slovnik_3)
```

```python
pismena = ["a", "b" ,"c", "d"]
vyskyt = [1, 10, 5, 0]
muj_zip = zip(pismena, vyskyt)
muj_slovnik_4 = dict(zip(pismena, vyskyt))
```

```python
print(muj_slovnik_4)
```

```python
pismena = ["a", "b" ,"c", "d"]
vyskyt = [1, 10, 5, 0]
muj_slovnik_99 = {}.fromkeys(pismena, vyskyt)
print(muj_slovnik_99)
```

```python
seznam_tuplu = [("Matous", 1), ("Lukas", 10), ("Jan", 13)]
muj_slovnik_5 = dict(seznam_tuplu)
```

```python
print(muj_slovnik_5)
```
