### **Ukázka #01**

```python
x = "nejaky_string"
y = "nejaky" + "_" + "string"
x is y
```

<details>
  <summary>❓Vysvětlení ❓</summary>

`True`


1. x = "nejaky_string":

* Tento řádek přiřadí řetězec "nejaky_string" do proměnné x. Představte si to jako nalepení štítku "x" na krabici, která obsahuje řetězec "nejaky_string".

2. y = "nejaky" + "_" + "string":

* Tento řádek vytvoří řetězec "nejaky_string" spojením (konkatenací) řetězců "nejaky", "_" a "string". Výsledek je pak přiřazen do proměnné y. Takže další krabice s označením "y" nyní také obsahuje "nejaky_string".

3. x is y:

* Tento řádek používá operátor is k porovnání identit x a y. Ptá se, zda x a y odkazují na tentýž objekt v paměti počítače, nejen zda se jejich hodnoty shodují.

V tomto specifickém scénáři, díky optimalizaci Pythonu zvané *string interning*, x a y budou pravděpodobně odkazovat na stejný objekt řetězce v paměti. Proto x is y by se vyhodnotilo jako True. Důvodem je, že Python se snaží ukládat pouze jednu kopii identických řetězců, aby šetřil paměť.

Chceš vědět víc? Zkus to třeba [tady](https://medium.com/techtofreedom/string-interning-in-python-a-hidden-gem-that-makes-your-code-faster-9be71c7a5f3e).

<details>
  <summary>❓Vysvětlení ❓</summary>

<br>

1. `o = "Matouš"`: Tento řádek přiřadí řetězec "Matouš" do proměnné `o`. Představte si to jako nalepení štítku "o" na krabici obsahující řetězec "Matouš".

2. `p = "Matouš"`: Tento řádek dělá totéž pro proměnnou `p` - přiřadí jí stejný řetězec "Matouš". Nyní máme další krabici označenou "p", která také obsahuje "Matouš".

3. `o is p`: Toto je jádro úryvku. Používá operátor is k porovnání identit `o` a `p`.

* Operátor is kontroluje, zda `o` a `p` odkazují na stejný objekt v paměti počítače, nikoli jen zda se jejich hodnoty shodují.

<br>

Proč by mohlo být `False`:

* V předchozím příkladu (s x a y) Python použil techniku ​​nazývanou `string interning` k uložení pouze jedné kopie řetězce "nejaky_string" v paměti, takže x i y na ni odkazovaly.

* String interning se však obvykle používá pro řetězce obsahující **pouze písmena ASCII, čísla a podtržítka**.

* Řetězec "Matouš" obsahuje znak 'š', který je mimo základní sadu ASCII.

* Z tohoto důvodu může Python vytvořit v paměti samostatné objekty pro `o` a `p`, i když obsahují stejnou hodnotu. Proto by se `o is p` v tomto případě vyhodnotilo jako False.

* V podstatě tento kód demonstruje, že i když dvě proměnné mohou obsahovat stejnou hodnotu řetězce, nemusí vždy odkazovat na stejné místo v paměti, zvláště pokud řetězec obsahuje znaky mimo základní sadu ASCII.

### **Ukázka #05**

```python
a = 256
b = 256
a is b
```

```python
a = 257
b = 257
a is b
```

<details>
  <summary>❓Vysvětlení ❓</summary>
    
Tento kód demonstruje jemné chování Pythonu, které se týká způsobu, jakým ukládá celá čísla v paměti.

1. `a = 257`: Tento řádek přiřadí celočíselnou hodnotu 257 do proměnné `a`.

2. `b = 257`: Tento řádek podobně přiřadí stejnou celočíselnou hodnotu 257 do proměnné `b`.

3. `a is b`: Toto je klíčová část. Operátor is kontroluje, zda `a` a `b` odkazují na stejný objekt v paměti. Nejde jen o porovnání jejich hodnot; kontroluje se, zda ukazují na stejné místo v paměti počítače.

<br>

Proč by to mohlo být `False`:

**Interning celých čísel**: Python má optimalizaci zvanou interning celých čísel. Aby se šetřila paměť, předběžně alokuje rozsah běžně používaných celých čísel (obvykle -5 až 256). Když použijete celé číslo v tomto rozsahu, Python znovu použije existující objekt v paměti, místo aby vytvářel nový.

Mimo rozsah: Pro celá čísla mimo tento předběžně alokovaný rozsah (jako je 257 v tomto případě) však Python vytváří v paměti samostatné objekty, i když mají stejnou hodnotu.

Výsledek: Proto se `a is b` v tomto scénáři pravděpodobně vyhodnotí jako False. To proto, že `a` a ``b, ačkoli obsahují stejnou hodnotu (257), pravděpodobně ukazují na různé objekty v paměti.

</details>

### **Ukázka #06**

```python
c = []
d = []
c is d
```

```python
c = tuple()
d = tuple()
c is d
```

<details>
  <summary>❓Vysvětlení ❓</summary>
    
Proč by to mohlo být `True`?

**Neměnnost a optimalizace**: N-tice jsou neměnné, což znamená, že je nelze po vytvoření změnit. Python často optimalizuje ukládání neměnných objektů. Když vytvoříte dvě prázdné n-tice, jako je tato, Python může v paměti vytvořit pouze jeden objekt n-tice a nechat na něj odkazovat `c` i `d`. To je podobné konceptu interningu řetězců, který jsme viděli dříve, ale pro n-tice.

Úspora paměti: Tato optimalizace šetří paměť, protože identické neměnné objekty nemusí být ukládány vícekrát.

V podstatě: Kód ukazuje, že i když vytvoříte dvě samostatné proměnné n-tice, pokud jsou identické a neměnné, Python je může pro efektivitu odkazovat na tentýž objekt v paměti. Operátor is by pak vrátil True, což by znamenalo, že sdílejí stejnou identitu.
</details>

### **Ukázka #08**

```python
True and True or False and not True
```

<details>
  <summary>❓Vysvětlení ❓</summary>

Tento kód používá booleovskou logiku a prioritu operátorů k určení konečného výsledku. Zde je postup krok za krokem:

1. `not True`: Operátor not invertuje pravdivostní hodnotu True, což vede k False.

2. `False and not True`: Toto se stane False and False, což se vyhodnotí jako False, protože oba operandy operátoru and musí být True, aby byl výraz True.

3. `True and True`: Toto se vyhodnotí jako True, protože oba operandy operátoru and jsou True.

4. `True and True or False and not True`: Nyní se výraz stane True or False, protože jsme dosadili výsledky z kroků 2 a 3.

5. `True or False`: Toto se vyhodnotí jako True, protože operátor or potřebuje pouze jeden ze svých operandů, aby byl True, aby byl celý výraz True.

<br>

Shrnutí:

Kód vyhodnocuje sérii logických výrazů pomocí operátorů and, or a not. Vzhledem k prioritě operátorů a způsobu, jakým tyto operátory fungují, je konečný výsledek True. Můžete si to představit takto:

* Nejprve se vyhodnotí not True a nahradí se False
* Poté se provedou dvě operace and (True and True a False and False)
* Tyto výsledky se použijí v operaci or (True or False)
* Operace or se vyhodnotí jako True.

</details>
