<br>

## **Kombinace s podmínkou**

---

klasicky:

````python
jmeno = input("JMENO: ")

if jmeno == "Lucie":
    print(f"POZDRAV: Ahoj, {jmeno}!")
else:
    print("POZDRAV: Ahoj, vsem!")
````

s walrusem

````python
if (jmeno := input("JMENO: ")) == "Lucie":
    print(f"POZDRAV: Ahoj, {jmeno}!")
else:
    print("POZDRAV: Ahoj, vsem!")
````

<br>

## **Kombinace s cyklem while**

---

````python
while (vstup := input("Zadej vstup ")) != "quit":
    print(f"Zadal jsi: {vstup}")
else:
    print("Konec smycky")
````
