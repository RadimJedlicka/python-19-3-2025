## **Smyčka WHILE**

---

````python
podminka = 1
````

````python
while podminka < 10:
    print(f'V tomto cyklu je {podminka=}, coz je mensi nez 10.')
    podminka = podminka + 1

print(f'Ted uz ale {podminka=} neni mensi nez 10')

print('Kod pokracuje pod smyckou.')
````

<br>


* `while` je klíčový výraz v záhlaví (zápisu) cyklu,
* `podminka < 10` je podmínka. Pokud je vyhodnocená jako `True`, proveď další krok; pokud `False`, pokračuj se zápisem pod
* řádek s předpisem musí být zakončený dvojtečkou
* následují odsazené instrukce, které se budou opakovat v každém kroku

## **WHILE loop s doplňující podmínkou**

---

Pokud je potřeba dopsat ještě doplňující podmínku, umí i `while` pracovat s těmi doplňky (stejně jako cyklus `for`):

````python
index = 0
````

````python
while index < 20:
    if len(str(index)) != 2:
        index = index + 1
        continue
    print(f"index={index}")
    index += 1
````

*Poznámka:* Opatrně, je rozumné zkusit se nad zápisem zamyslet, jestli neexistuje možnost, jak podmínku upravit přímo v předpisu

````python
index = 10
````

````python
while index < 20 and len(str(index)) == 2:
    print(f"index={index}")
    index += 1
````

## **klauzule `else`**

---

````python
cislo = 5

while cislo > 0:
    cislo -= 1
    print(cislo)
else:
    print('Loop ended.')
````

````python
cislo = 5

while cislo > 0:
    cislo -= 1
    print(cislo)
    if cislo == 2:
        break
    else:
        print('Ahoj')
else:
    print('Loop ended.')
````

Pokud ve smyčce použiji `break`, nedostanu se na klauzuli `else`

<br>

*poznámka*. větev `else` se u smyčky typu `while` liší od smyčky typu `for`. Pokud bude vyhodnocena podmínka v předpisu jako nepravdivá (`False`), proveď zápis v `else` a pokračuj pod smyčkou.

<br>

U cyklu `for` fungovala větev `else` na principu toho, že ke spuštění dojde tehdy, pokud cyklus provedl bez přerušení svůj předpis.

<br>

Nekonečný cyklus s pomocí `while` je samozřejmě koncept, kterého se běžně využívá, pokud je podmínka a průběh validní:

````python
while True:
    vstupni_hodnota = input("Zadej pismeno: ")

    if vstupni_hodnota.isalpha():
        print(f"'{vstupni_hodnota}' je opravdu pismeno.")
        print()
    else:
        print(f"'{vstupni_hodnota}' neni pismeno,  koncim!")
        break
````

bez `break`

````python
is_true = True

while is_true:
    vstupni_hodnota = input("Zadej pismeno: ")

    if vstupni_hodnota.isalpha():
        print(f"'{vstupni_hodnota}' je opravdu pismeno.")
        print()
    else:
        print(f"'{vstupni_hodnota}' neni pismeno,  koncim!")
        is_true = False
else:
    print('Bez se doucit pismena!')
````

#### **1. Počet výskytů**

Metoda `.get` je velmi užitečná, když potřebujeme spočítat počet výskytů prvků. Abychom pochopili následující kód, musíme vědět něco o smyčce while.

Můžeme postupovat tímto způsobem:

````python
barvy = [
    'zelena', 'modra', 'cerna', 'cervena',
    'cervena', 'zluta', 'modra', 'seda',
    'cerna' , 'cervena', 'zelena'
    ]

pocet_barev = {}

while barvy:
    barva = barvy.pop()

    if barva not in pocet_barev:
        pocet_barev[barva] = 0

    pocet_barev[barva] = pocet_barev[barva] + 1
````

````python
pocet_barev
````

Efektivnější je ale tento postup:

````python
barvy = [
    'zelena', 'modra', 'cerna', 'cervena',
    'cervena', 'zluta', 'modra', 'seda',
    'cerna' , 'cervena', 'zelena'
    ]

pocet_barev = {}

while barvy:
    barva = barvy.pop()
    pocet_barev[barva] = pocet_barev.get(barva,0) + 1
````

````python
pocet_barev
````

#### **2. Předem neznáme počet opakování**

Smyčky while jsou často používány v případech, kdy **předem neznáme počet opakování**, který má být vykonán. Například když program vyžaduje vstup od uživatele.

Program níže požaduje číslo a poté odpočítává jeden cyklus za sekundu.

````python
from time import sleep

pocet_sekund = int(input('Kolik sekund potrebujes?: '))

while pocet_sekund:
    sleep(1)
    print(pocet_sekund)
    pocet_sekund = pocet_sekund - 1
````

Co bude výstupem následujícího kusu kódu?

````python
mesta = {'Praha': 1, 'Brno': 2, 'Ostrava': 3}
while mesta:
    print(mesta.popitem())
print('Hotovo.')
````

<br>

Co bude výstupem následujícího kusu kódu?

````python
mesta = {'Praha': 1, 'Brno': 2, 'Ostrava': 3}
while len(mesta) > 3:
    print(mesta.popitem())
print('Hotovo.')
````

<br>

Co bude výstupem následujícího kusu kódu?

````python
mesta = ['Praha', 'Brno', 'Ostrava', 'Plzen', 'Liberec']
while mesta:
    if len(mesta) < 3:
        break
    print(mesta.pop())
print('Hotovo.')
````

<br>

Tento kód je stejný jako v předchozím případu, ale místo `break` je zde ohlášení `continue`.
Co bude výstupem tohoto kusu kódu?


````python
mesta = ['Praha', 'Brno', 'Ostrava', 'Plzen', 'Liberec']
while mesta:
    if len(mesta) < 3:
        continue
    print(mesta.pop())
print('Hotovo.')
````

Jaká bude hodnota proměnné `vysledek`, když spustíme následující kód?

````python
vysledek = ""

n = 5
while n > 0:
    n -= 1
    if (n % 2) == 0:
        continue

    mesta = ['Praha', 'Brno', 'Ostrava']
    while mesta:
        vysledek += str(n) + mesta.pop(0)
        if len(mesta) < 2:
            break
````
