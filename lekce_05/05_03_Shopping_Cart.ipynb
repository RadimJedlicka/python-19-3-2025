## **Zadání**

Cílem této úlohy je nechat uživatele přidávat do proměnné `kosik` tak dlouho, dokud bude chtít. Celá situace bude představovat proces virtuálního nákupu.

<br>

Celá úloha se bude skládat z těchto kroků:

1. Pozdravit uživatele a vypsat mu aktuální nabídku (proměnná `nabidka`),
2. zapsat cyklus, který umožní postupně zadávat hodnoty do košíku (proměnná `kosik`) tak dlouho, dokud si to bude uživatel přát,
3. pokud uživatel zadá vstup do proměnné `zbozi` takové zboží, které není v proměnné `sklad`, vypiš: `"Zboží <zbozi> není v nabídce"`,
4. pokud zboží v proměnné `kosik` není, přidej 1 kus do košíku a odeber 1 kus z proměnné `sklad` (index 1),
5. pokud zboží v proměnné `kosik` je, uprav původní hodnotu +1 kus a z proměnné `sklad` opět odeber 1 kus,
6. pokud se v proměnné `sklad`, na indexu 1 objeví u konkrétní sklad hodnota 0, potom vypiš:`"Zboží <zbozi> není skladem"`,
7. na závěr vypiš pomocí funkce `print` obsah proměnné `kosik`.

## **Vstupní údaje, které použijeme**

````python
kosik = dict()
oddelovac = "=" * 40
sklad = {
    "mleko": [30, 5],  # index '0'-> cena, index '1' -> pocet
    "maso": [100, 1],
    "banan": [30, 10],
    "jogurt": [10, 5],
    "chleb": [20, 5],
    "jablko": [10, 10],
    "pomeranc": [15, 10]
}
nabidka = """
+-----------+----------+
| POTRAVINA |   CENA   |
+-----------+----------+
| mleko     |    30,-  |
| maso      |   100,-  |
| banan     |    30,-  |
| jogurt    |    10,-  |
| chleb     |    20,-  |
| jablko    |    10,-  |
| pomeranc  |    15,-  |
+-----------+----------+
"""
````

## **Zapsat cyklus, který umožní postupně zadávat hodnoty**

do virt. košíku (proměnná kosik) tak dlouho, dokud si to bude uživatel přát

## **A jdeme do ➰ smyčky:**

* ### Pokud uživatel zadá vstup do proměnné `zbozi`:

    👉 zboží, které není ve slovníku `sklad`, vypiš: "Zboží není v nabídce"
    
    <font color=ligtblue><b> 💻 Kód:</b>

````python
if zbozi not in sklad:
    print(f'{zbozi.title()} bohuzel neni v nabidce')
````

#  
* ### Pokud zboží v proměnné `kosik` není:

    👉 přidej 1 kus do košíku a odeber 1 kus z proměnné sklad (index 1)
    
    <font color=ligtblue><b> 💻 Kód:</b>

````python
elif zbozi not in kosik and sklad[zbozi][1] > 0:
        kosik[zbozi] = [sklad[zbozi][0], 1]
        sklad[zbozi][1] = sklad[zbozi][1] - 1
````

#  
* ### Pokud zboží v proměnné `kosik` je:

    👉 uprav původní hodnotu +1 kus a z proměnné sklad opět odeber 1 kus
    
    <font color=ligtblue><b> 💻 Kód:</b>

````python
elif zbozi in kosik and sklad[zbozi][1] > 0:
        kosik[zbozi][1] = kosik[zbozi][1] + 1
        sklad[zbozi][1] = sklad[zbozi][1] - 1
````

#  
* ### Pokud se v proměnné `sklad`, na indexu `1` objeví u konkrétní sklad hodnota `0`:

    👉 potom vypiš: "Zboží {zbozi} není skladem"
    
    <font color=ligtblue><b> 💻 Kód:</b>

````python
elif sklad[zbozi][1] == 0:
        print(f"Zbozi '{zbozi}' neni skladem")
````

    👉 potom vypiš: "Zboží {zbozi} není skladem"

### <font color=ligtblue><b> 💻 Kompletní kód:</b>

````python
# Zadané hodnoty
kosik = dict()
oddelovac = "=" * 40
sklad = {
    "mleko": [30, 5],  # index '0'-> cena, index '1' -> pocet
    "maso": [100, 1],
    "banan": [30, 10],
    "jogurt": [10, 5],
    "chleb": [20, 5],
    "jablko": [10, 10],
    "pomeranc": [15, 10]
}
nabidka = """
+-----------+----------+
| POTRAVINA |   CENA   |
+-----------+----------+
| mleko     |    30,-  |
| maso      |   100,-  |
| banan     |    30,-  |
| jogurt    |    10,-  |
| chleb     |    20,-  |
| jablko    |    10,-  |
| pomenrac  |    15,-  |
+-----------+----------+
"""

# Úvodní sekce s uvítáním a výpisem zboží
print(
    "Vitejte u naseho nakupniho kosiku!".center(len(oddelovac)),
    oddelovac,
    nabidka,
    sep="\n"
)

# Cyklus, který přidává zboží do košíku tak dlouho, jak uživatel zamýšlí
#while (zbozi := input("Zbozi:")) != "q": # walrus nefunkcni v Colabu
prepinac = True

while prepinac:
    select = (input('Vlozit do kosiku: '))
    if select == 'q':
        prepinac = False

    elif select not in sklad:
        print(f'{select} neni v nabidce')

    elif select not in kosik and sklad[select][1] > 0:
        kosik[select] = [sklad[select][0], 1]
        sklad[select][1] -= 1

    elif select in kosik and sklad[select][1] > 0:
        kosik[select][1] += 1
        sklad[select][1] -= 1

    elif sklad[select][1] == 0:
        print('Toto zbozi jiz neni skladem')

# Finální výpis košíku po nákupu
else:
    print(oddelovac, kosik, sep="\n")


````
